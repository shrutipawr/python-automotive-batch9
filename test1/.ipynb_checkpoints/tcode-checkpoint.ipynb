```python
def check_eligibility(age, income, loan):
    if age <= 0 or income < 0 or loan < 0:
        return "Invalid Input"
    elif age < 21 or age > 60 :
        return "Not Eligible"
    elif income < 20000:
        return "Not Eligible"
    elif loan > 0.4 * income:
        return "Eligible"
    else:
        return "Eligible"
        
```
